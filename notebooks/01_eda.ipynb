```python
reference_date = df["InvoiceDate"].max()

print("Reference Date:", reference_date)
```

```python
rfm = df.groupby("CustomerID").agg({
    "InvoiceDate": lambda x: (reference_date - x.max()).days,
    "InvoiceNo": "nunique",
    "Revenue": "sum"
})

rfm.columns = ["Recency", "Frequency", "Monetary"]

rfm.head()
```

```python
fig = px.histogram(
    rfm,
    x="Recency",
    nbins=50,
    title="Distribution of Customer Recency",
    labels={"Recency": "Recency (Days)"}
)

fig.show()
```

```python
fig = px.histogram(
    rfm,
    x="Frequency",
    nbins=50,
    title="Distribution of Customer Frequency",
    labels={"Frequency": "Number of Purchases"}
)

fig.show()



```

```python
fig = px.histogram(
    rfm,
    x="Monetary",
    nbins=50,
    title="Distribution of Customer Monetary Value",
    labels={"Monetary": "Total Spending"}
)

fig.show()
```

```python
fig = px.scatter(
    rfm,
    x="Frequency",
    y="Monetary",
    hover_name=rfm.index.astype(str),
    title="Customer Frequency vs Monetary Value",
    labels={
        "Frequency": "Purchase Frequency",
        "Monetary": "Total Spending"
    }
)

fig.show()
```
